```python
import pickle, math
data = pickle.load(open('/home/tsuruoka/hdd/BEV/CarlaRunner/DrivingAgent/notebooks/tmp/_carla_infer.pkl','rb'))
info = data['infos'][0]
quat = info['ego2global_rotation']
print(info["lidar2ego_translation"])
yaw_deg = math.degrees(math.atan2(2*(quat[0]*quat[3]+quat[1]*quat[2]),
                                  1-2*(quat[2]**2+quat[3]**2)))
print('ego yaw', yaw_deg)
print('can_bus yaw', info['can_bus'][-1] if info['can_bus'] else None)
```
